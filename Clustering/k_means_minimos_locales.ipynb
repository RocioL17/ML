```python
# 🟣 PROBLEMA DE MÍNIMOS LOCALES EN K-MEANS

from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# Datos
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.5, random_state=0)

# Probar distintas inicializaciones
resultados = []
for n_init in [1, 5, 10, 20]:
    kmeans = KMeans(n_clusters=4, n_init=n_init, random_state=0)
    kmeans.fit(X)
    resultados.append(kmeans.inertia_)

# Mostrar resultados
plt.plot([1,5,10,20], resultados, marker='o')
plt.xlabel("Número de inicializaciones (n_init)")
plt.ylabel("Inercia final")
plt.title("Impacto de n_init en K-Means")
plt.show()
```
